```python
import numpy as np
import matplotlib.pyplot as plt
import sunpy.map

%matplotlib widget
```

```python
hmi_hr_dir = 'E:/Research/Data/SDO/HMI/Magnetogram/hr/'
hmi_hr_file = 'hmi.M_720s.20240504_150000_TAI.fits'
hmi_hr_path = hmi_hr_dir + hmi_hr_file

hmi_hr_map = sunpy.map.Map(hmi_hr_path)
```

```python
hmi_dir = 'E:/Research/Data/SDO/HMI/Magnetogram/720s/'
hmi_file = 'hmi.M_720s.20240504_143600_TAI.3.magnetogram.fits'
hmi_path = hmi_dir + hmi_file

hmi_map = sunpy.map.Map(hmi_path)
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection=hmi_hr_map)
im = hmi_hr_map.plot()
plt.colorbar(im, ax=ax, label=hmi_hr_map.unit.to_string())
```

```python
hmi_map_rotated = hmi_map.rotate(order=3)

fig = plt.figure()
ax = fig.add_subplot(projection=hmi_map_rotated)
im = hmi_map_rotated.plot()
plt.colorbar(im, ax=ax, label=hmi_map_rotated.unit.to_string())
```

```python
hmi_Br = hmi_map_rotated.data

plt.figure()
# hmi_Br_abs_max = np.nanmax(np.abs(hmi_Br))
hmi_Br_abs_max = 3300
plt.pcolormesh(hmi_Br, cmap='gray', vmin=-hmi_Br_abs_max, vmax=hmi_Br_abs_max)
plt.colorbar()
plt.axis('equal')
plt.show()
```
